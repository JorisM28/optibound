```python
def euclidian_distance(X, Y):
    return torch.linalg.vector_norm((X - Y).float(), dim=-1)
```

```python
X = torch.tensor([
    [1, 2],
    [6, 5],
    [3, 8]
])
```

```python
Y = torch.tensor([
    [5, 6], 
    [7, 1],
    [3, 2]
])
```

```python
Y
```

```python
import numpy as np
D = euclidian_distance(X, Y)
```

```python
def browse_and_bin(X, epsilon):

    X_flat = X.flatten()
    
    masque = X_flat >= epsilon
    
    return X_flat[masque]
   
```
